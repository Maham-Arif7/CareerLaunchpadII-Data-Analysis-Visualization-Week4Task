## 2. Data Import & Cleansing

###Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

###Read the Data file

```python
df = pd.read_csv("/content/data.csv", sep='\t')
df.head()
```

###Explore Data

```python
df.head(5) # Display first 5 rows
df.shape # Check data dimensions
df.isnull().sum() # Check for missing values
```

###Handle Missing values,identify mode of the column with missing values

```python
a =df["When do you prefer to study?"].mode()
a[0] # Access the mode valu
```

###Fill missing values with the mode

```python
df.fillna(a[0], inplace=True)
df.isnull().sum() # Verify missing values are filled
```

## 4. Data Visualization

### Univariate Analysis

#### Age distribution

```python
plt.figure(figsize=(6, 4))
age_counts = df['What is your age?'].value_counts()
plt.bar(age_counts.index, age_counts.values, color='teal')
plt.title('Distribution of Age')
plt.xlabel('Age Groups')
plt.ylabel('Number of Students')
plt.show()
```

####  Pie Chart to show proportions

```python
age_counts = df['What is your age?'].value_counts()
plt.pie(age_counts, labels=age_counts.index, autopct='%1.1f%%')
plt.title("Distribution of Age")
plt.show()
```

####  Daily Screen Time

```python
screen_time_counts = df["How many hours do you spend on screens daily? (including phone, laptop, tablets )"].value_counts()
plt.bar(screen_time_counts.index, screen_time_counts.values, color='khaki')
plt.title("Daily Screen Time")
plt.xlabel("Hours")
plt.xticks(rotation=45)
plt.ylabel("Count")
```

#### Coffee or Tea Preference

```python
coffee_tea_counts = df["Do you prefer coffee or tea?"].value_counts()
plt.pie(coffee_tea_counts, labels=coffee_tea_counts.index, autopct='%1.1f%%',
colors=['orange', 'skyblue', 'teal'])
plt.title("Coffee or Tea Preference")
plt.show()
```

####  Preferred Study Time

```python
study_time_counts = df["When do you prefer to study?"].value_counts()
plt.bar(study_time_counts.index, study_time_counts.values)
plt.title("Preferred Study Time")
plt.xlabel("Time of Day")
plt.ylabel("Count")
plt.show()
```

####  Which social media app do you prefer?

```python
social_media_counts = df['Which social media app do you prefer?'].value_counts()
plt.bar(social_media_counts.index, social_media_counts)
plt.title('Social Media App Preferences')
plt.xlabel('Social Media App Categories')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)
plt.show()
```

#### What are your hobbies?

```python
hobbies_media_counts = df['What are your hobbies / interest?'].value_counts()
plt.barh(hobbies_media_counts.index, hobbies_media_counts)
plt.title('Hobbies / interest')
plt.xlabel('Hobbies')
plt.ylabel('Number of Students')
plt.show()
```

#### Exercise Frequency per Week

### Bivariate Analysis

####  Age vs.Preferred Study Time

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='What is your age?', hue='When do you prefer to study?', data=df)
plt.title('Age vs. Preferred Study Time')
plt.tight_layout()
```

#### Age vs.Exercise Frequency

```python
sns.countplot(x='What is your age?', hue='How often do you exercise in a week?', data=df)
plt.title('Age vs. Exercise')
plt.tight_layout()
```

####  Daily Screen Time vs. Age

```python
sns.countplot(x='How many hours do you spend on screens daily? (including phone, laptop, tablets )', hue='What is your age?', data=df)
plt.title('Screen Time vs. Age')
plt.xticks(rotation=45)
plt.tight_layout()
```

####  Coffee or Tea Preference vs. Number of Languages Spoken

```python
sns.countplot(x='How many languages do you speak fluently?', hue='Do you prefer coffee or tea?', data=df)
plt.title('Coffee or Tea Preference vs. Number of Languages Spoken')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

####  Commute Method vs.Exercise Frequency

```python
plt.figure(figsize=(10, 5))
sns.countplot(x='How do you primarily commute to work or school?', hue='How often do you exercise in a week?', data=df,width=0.6)
plt.title('Commute Method vs. Exercise Frequency')
plt.xlabel('Count')
plt.ylabel('Commute Method')
plt.legend(title='Exercise Frequency',fontsize='small')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Sleep Hours vs. Daily Screen Time

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='How many hours of sleep do you get on an average night?', hue='How many hours do you spend on screens daily? (including phone, laptop, tablets )', data=df,width=0.7)
plt.title('Sleep Hours vs. Daily Screen Time')
plt.xlabel('Sleep Hours')
plt.ylabel('Count')
plt.legend(title='Screen Time', loc='upper right')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
